```python
num = int(input("Enter the number: "))

original_num = num
reversed_num = 0

if(num < 0):
    print("Not Palindrome")
else:
    while num != 0:
        singleDigit = num % 10
        reversed_num = reversed_num * 10 + singleDigit
        num = num // 10
    
    # print("Reversed: ", reversed_num)
            
    if(original_num == reversed_num):
        print("Palindrome")
    else:
        print("Not Palindrome")

```
