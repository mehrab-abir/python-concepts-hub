Get the text from the sample.txt file, and count the following:
1. Number of lines
2. Number of words
3. Number of characters

```python
with open('./sample.txt','r') as file:
    total_lines = len(file.readlines())
    
    file.seek(0)
    
    total_words = 0
    total_characters = 0
    
    for line in file:
        words_in_line = len(line.strip().split())
        total_words = total_words + words_in_line
        # print(line.strip().split())
        
        lineArr = line.strip().split()
        for word in lineArr:
            total_characters = total_characters + len(word)
    
    print("-----------------------------")
    print("Total lines: ", total_lines)    
    print("Total words : ", total_words)
    print("Total charcaters : ", total_characters) 
    
# storing those info into another file
with open('./sample_file_metaData.txt','w') as file:
    file.write(f"Total lines: {total_lines}\n")
    file.write(f"Total words: {total_words}\n")
    file.write(f"Total characters: {total_characters}")

print("File with all counts created")
    
    
```
