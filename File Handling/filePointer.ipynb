```python
with open('./ChildFiles/maple.txt','r') as file:
    print(f"Cursor at index : {file.tell()}") # tells where the cursor is currently
    
    print(file.read(11)) # read first 11 charcters (0-10), cursor stops at 12th index
    
    print(f"Cursor at index : {file.tell()}") # tells where the cursor is currently
    
    # read from index 12 to untill end
    print(file.read())
    
    print(f"Cursor at index : {file.tell()}")
```

```python
with open('./ChildFiles/maple.txt','r') as file:
    print(f"Cursor at index : {file.tell()}") # tells where the cursor is currently
    
    print(file.read(11)) # read first 11 charcters (0-10), cursor stops at 12th index
    
    print(f"Cursor at index : {file.tell()}") # tells where the cursor is currently
    
    print('---------------')
    file.seek(0) # cursor goes back to 0
    
    print("After cursor taken to 0 index:")
    print(file.read(11)) # read from beginning again..index (0 - 10)
```

```python
with open('./ChildFiles/maple.txt','r') as file:
    print(f"Cursor at: {file.tell()}") # index 0
    
    file.seek(12) # cursor moves to index 12
    print(f"Cursor at: {file.tell()}") # index 12
    
    print("\nReading from index 12:")
    print(file.read()) # starts reading from index 12
```
