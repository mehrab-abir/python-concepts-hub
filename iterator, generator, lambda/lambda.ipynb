#### With map()

````python
nums = [1,2,3,4,5]

squared = list(map(lambda x: x*x, nums))

print(squared)
````

#### with filter

````python
nums = [9,6,3,1,5,7,9,4,2]

even = list(filter(lambda x : x % 2== 0, nums))

print(even)
````
