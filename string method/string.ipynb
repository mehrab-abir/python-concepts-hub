### Formatting 


```python
height = 5.13566 

print(f"Height: {height:.2}") # total 2 digits -->> 5.1
print(f"Height: {height:.3}") # total 3 digits -->> 5.14
```

```python
radius = 6

area = 3.1415926534897932 * radius * radius
print(f"Area : {area:.7}") # Area : 113.0973
```
