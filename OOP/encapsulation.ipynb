```python
class Account:
    def __init__(self, accName, accNumber, balance):
        self.__accName = accName; # Private member
        self.accNumber = accNumber;
        self.__balance = balance; # Private member
        
    def deposit(self, amount):
        if(amount <= 0):
            print("Invalid amount")
        else:
            self.__balance = self.__balance + amount;
    
    def withdraw(self, amount):
        if(amount <= 0 or amount > self.__balance):
            print("Invalid amount")
        else:
            self.__balance = self.__balance - amount;
            
    def getBalance(self):
        return self.__balance
    
    def getName(self):
        return self.__accName

    def setName(self,name):
        self.__accName = name;
    
jhon_account = Account("Jhon Doe","100004",5000)

print(jhon_account.getBalance())
# print(jhon_account.__balance) # private members can;t be access from outside

jhon_account.deposit(2000);

print("After deposit:")
print(jhon_account.getBalance())

jhon_account.withdraw(500);
print("After withdraw:")
print(jhon_account.getBalance())

jhon_account.setName("JHON_ACC")

print(jhon_account.getName())

# jhon_account.deposit(-153) # invalid amount
# jhon_account.withdraw(8745632) # invalid amount
# jhon_account.withdraw(-122) # invalid amount
    
```
