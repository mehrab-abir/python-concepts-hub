```python
class Camera:
    def __init__(self,name):
        self.name = name
    
    def capture(self):
        print("Photo is captured")
    
class SmartPhone(Camera):
    def __init__(self, name, resolution):
        super().__init__(name)
        self.resolution = resolution
    
    #override
    def capture(self):
        print("Photo is captured by smart phone")
        
class DSLR(Camera):
    def __init__(self, name, resolution):
        super().__init__(name)
        self.resolution = resolution
    
    # override
    def capture(self):
        print("Photo is captured by dslr")

class Drone(Camera):
    def __init__(self, name, resolution):
        super().__init__(name)
        self.resolution = resolution
    
    # override
    def capture(self):
        print("Photo is captured by drone")
        
smartPhone = SmartPhone("sm",100)
dslr = DSLR("dslr",200)
drone = Drone("drone",150)

smartPhone.capture()
dslr.capture()
drone.capture()
```
