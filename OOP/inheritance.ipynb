```python
class Phone:
    def __init__(self, brand, release_year, battery):
        self.brand = brand;
        self.release_year = release_year;
        self.battery = battery;
    
    def turn_on(self):
        print("Phone turned on")

nokia = Phone("Nokia",1995,2000)

nokia.turn_on()
print(f"Brand: {nokia.brand}\nRelease Year: {nokia.release_year}\nBattery: {nokia.battery}")        
```

```python
class SmartPhone(Phone):
    def __init__(self, brand, release_year, battery, camera, charge=100):
        super().__init__(brand, release_year, battery)
        self.camera = camera;
        self.charge = charge;
    
    def capture_photo(self,number):
        self.charge = self.charge - number
        print(f"{number} Photo captured")

apple = SmartPhone("Apple","2025","4000mph","30 MP",95)

print(f"Charge Now: {apple.charge}")

apple.capture_photo(2)

print(f"Charge Now: {apple.charge}")

print("Camera: ",apple.camera)

print("Release year: ",apple.release_year)
print("Battery: ",apple.battery)
```

Multiple Inheritance

```python
# a different class
class AIFeature:
    def __init__(self,ai_model):
        self.ai_model = ai_model
        
    def turn_on_ai_mode(self):
        print("AI Mode is on now!")
```

```python
# Multiple inheritance

class HiTechPhone(SmartPhone, AIFeature):
    def __init__(self, brand, release_year, battery, camera, charge,ai_model):
        SmartPhone.__init__(self, brand, release_year, battery, camera, charge)
        AIFeature.__init__(self, ai_model)

ProTechPhone = HiTechPhone("XYZ","2026","6000mph","40 MP",100,"Neon AI")

ProTechPhone.turn_on() # SmartPhone(Phone) theke
print("Battery: ",ProTechPhone.battery) # SmartPhone(Phone) theke
print("Release: ",ProTechPhone.release_year) # SmartPhone(Phone) theke
print("Charge Now: ",ProTechPhone.charge) # SmartPhone theke
ProTechPhone.capture_photo(10) # SmartPhone theke
print("Charge Now: ",ProTechPhone.charge) # SmartPhone theke

print("------------------------")
ProTechPhone.turn_on_ai_mode() # AIFeature theke
print("AI Model: ",ProTechPhone.ai_model) # AIFeature theke
```
