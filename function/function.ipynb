*args


```python
def getSum(*args):
    print(f"args --> {args}")
    
    sum = 0;
    
    for i in args:
        sum =  sum + i
    
    return sum

result = getSum(1,3,8,5,4)

print(f"sum = {result}")
```

Keyword Arguments

```python
def f(**kwargs):
    # print(kwargs)
    
    for key,value in kwargs.items():
        print(f"{key}: {value}")
    
f(a=10,b=20,c=30) # returns a dictionary with this items
```

return type- multiple data return and unzipping

```python
def nums():
    return 10,25; # returing multiple data

print(type(nums())) # return type -> tuple

a,b = nums() # unzipping multiple data

print(f"a = {a}\nb = {b}")
```
