Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered*, changeable and do not allow duplicates.

```python
#dictionary,,like object in JS

Player = {
    "name" : "Neymar",
    "jersyNum" : 10,
    "country" : "Brazil",
    "position" : "Left Winger"
}

print(Player)
print('-------------')

print(Player["name"])

print(Player.get("position"))

Player["name"] = "Neymar Jr."

print(Player["name"])

del Player['jersyNum'] # delete a key 

print('-------------')
print(Player)

Player.pop("country") # delete a key - country

print('-------------')
print(Player)
```

```python
Student_one = {
    "name" : "Jhon Doe",
    "age" : 20,
}

print(Student_one.get("marks")) # none
print(Student_one.get("marks",0)) # 0

for x in Student_one:
  print(x)
  
print("........")
  
for x in Student_one:
  print(Student_one[x])

print("------------")

for x in Student_one.values():
  print(x)
```

```python
## this is correct, because 'keys' must be immutable, and here, tuple is immutable, so a tuple as a 'key' works here,,, just like 'string' as 'key' is used
dict_1 = {
    (10,20,30) : "some numbers in a tuple",
    "name" : "dictionary",
    #[3,1,4] : "a list of numbers"  ###--> this won't work as 'list' is mutable
}

print(dict_1)
```

```python
Inception = {
    "name" : "Inception",
    "year" : 2010,
    "genre" : "Sci-fi/Action"
}

incep_keys = Inception.keys()
incep_values = Inception.values()

print(f"Keys --> {incep_keys}\nValues --> {incep_values}")
print("-------------")

Inception["director"] = "Christopher Nolan" # incep_keys and incep_values will be updated automatically

print(f"Keys --> {incep_keys}\nValues --> {incep_values}")
valuesArr = list(incep_values)
print(f"Values array : {valuesArr}")

print("-------------")
print(Inception.items()) # key-value in a tuple, whole thing inside a list

print("-------------------")

# iteration
for key,value in Inception.items():
    print(f"{key} -> {value}")

```
