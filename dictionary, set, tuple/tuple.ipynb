Tuple - A built-in data structure used to store an ordered, immutable collection of items

```python
# immutable
tuple_one = (10,20,30,20,40,20)

print(tuple_one[1])
print(tuple_one)

print(tuple_one.count(20)) #--> 3

my_tuple = ("mango", 50, True) # tuple
print(my_tuple)


single_tuple = ("apple",)  # This is a tuple
not_a_tuple = ("apple")    # This is a string

```

Loop through

```python
thistuple = ("mango", "banana", "cherry")
for x in thistuple:
  print(x)
  
print(thistuple.index('cherry')) # --> 2
```

Join tuple

```python
tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```
