A set is a collection which is unordered, unchangeable*, and unindexed. Set items are unchangeable, but you can remove items and add new items.

```python
colorSet = {'blue','black','green','cyan'}

print(colorSet)

print('cyan' in colorSet)

colorSet.remove('black')

print(colorSet)

numSet = {5,6,3,1,4,5,2}

print(numSet) # sorted, no duplicate

numSet.add(56)

print(numSet)

for color in colorSet:
    print(color)
```

The union() method returns a new set with all items from both sets.

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)

# The | operator can be used instead of the union() method, and the result will be same.

set3 = set1 | set2
print(set3)
```

Join Multiple Sets

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1.union(set2, set3, set4) # or myset = set1 | set2 | set3 |set4
print(myset)
```

Update set - The update() method inserts the items in set2 into set1:

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}

set1.update(set2)
print(set1)
```

### Set Intersection
Return a set that contains the items that exist in both set x, and set y:

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.intersection(y)

print(z)
```
