### Sensor Calibrator — Min-Max Scaling
You’re calibrating a tiny sensor before training a model. Given x, min_v, and max_v, compute: </br>

norm = (x - min_v) / (max_v - min_v) </br>
Then print norm with 2 decimal places.

```python
x, min_v, max_v = map(float,input().split())

norm = (x - min_v) / (max_v - min_v)

print(f"{norm:.2f}")
```

### Mini Loss Tracker

Your tiny trainer reports a stream of loss values. Read how many values to process, then read that many losses and compute the average. If the average loss is ≤ target, print PASS; otherwise print RETRY.

```python
N = int(input("How many: "))
target = float(input("Target: "))

sum = 0;

for i in range(N):
    x = float(input())
    sum = sum + x

avg = sum / N

if(avg <= target):
    print("PASS")
else:
    print("RETRY")
```

### Majority Vote Gate
You’re simulating an ensemble of mini-models that each vote YES or NO on admitting a user. Read how many votes there are, then read each vote. If the number of YES votes is greater than or equal to the number of NO votes, print ACCEPT; otherwise print REJECT.

```python
n = int(input("Number of votes: "))
yes_count = 0
no_count = 0

for i in range(n):
    vote = input()
    if(vote == 'YES'):
        yes_count += 1
    else:
        no_count += 1

if(yes_count >= no_count):
    print("ACCEPT")
else:
    print("REJECT")
```

### AI Mode Detector
In real AI systems, mood detection uses natural language processing (NLP) to understand emotions from text. For now, let's simulate a mini version of that!</br>

You are given a sentence that represents a user's message. Your job is to guess if the mood is happy, sad, or neutral using simple logic: </br>

If the message contains words like happy, joy, or smile, print Happy Mood.
If it contains words like sad, cry, or angry, print Sad Mood.
Otherwise, print Neutral Mood.

```python
text = input("Enter the text: ").lower()

if any(word in text for word in ['happy','joy','smile']):
    print("Happy Mood")
elif any(word in text for word in ['sad','angry','cry']):
    print("Sad Mood")
else:
    print("Neutral Mood")
```

### AI Brightness Analyzer
AI systems often analyze images by looking at the brightness of pixels. Each pixel has a brightness value, usually between 0 (dark) and 255 (bright). Let's simulate this concept! </br>

You are given a list of integers representing pixel brightness values. Your task is to find the average brightness and decide whether the image is Dark, Normal, or Bright.</br>

If the average brightness < 85 → print Dark Image</br>
If 85 ≤ average brightness ≤ 170 → print Normal Image</br>
If average brightness > 170 → print Bright Image</br>

```python
pixels = list(map(int,input("Enter pixel values: ").split()))

total = 0

for value in pixels:
    total = total + value
    
avg = total/(len(pixels))

if(avg < 85):
    print("Dark Image")
elif(avg >= 85 and avg <= 170):
    print("Normal Image")
else:
    print("Bright Image")
```

### AI Bias Detector

In AI, bias means the model might favor one category more than others. Let’s simulate this idea with lists and conditional logic! </br>

You are given a list of predictions made by an AI model. Each prediction is either 'A' or 'B'.</br>

Your task is to check whether the model is biased or fair:</br>

If the count of 'A' or 'B' is more than 70% of all predictions, print Biased Model.
Otherwise, print Fair Model.

```python
predictons = list(map(str,input("Enter predictions: ").split()))

print(predictons)

total = len(predictons)

count_a = 0
count_b = 0

for value in predictons:
    if(value == 'A'):
        count_a += 1
    else:
        count_b += 1
        
verdict_a = (count_a * 100)/total
verdict_b = (count_b * 100)/total

if(verdict_a > 70 or verdict_b > 70):
    print("Biased Model")
else:
    print("Fair Model")
```

### AI Keyword Counter

AI systems often scan user messages to find keywords and understand intent. Let’s build a simple version of that! </br>

You are given a user's message and a list of AI-related keywords. Your task is to count how many of those keywords appear in the message.</br>

The keywords to look for are:</br>

ai, data, model, learn, train, neural</br>

If 2 or more keywords are found, print AI Detected. Otherwise, print Not AI Related.

```python
message = input("Enter the message: ").lower().split()

keywords = ['ai', 'data', 'model', 'learn', 'train', 'neural']

count = sum(1 for word in keywords if word in message)

if count >= 2:
    print("AI Detected")
else:
    print("Not AI Related")
```
