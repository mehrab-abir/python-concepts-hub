```python
nums = [x for x in range(100)]

# print(nums)

def data_loader(chunk_size, nums):
    for i in range(0,len(nums),chunk_size):
        yield nums[i:i+chunk_size]
        
x = data_loader(5,nums)

print(type(x)) # generator

print(next(x)) #[0, 1, 2, 3, 4] --> a small chunk
print(next(x)) #[5, 6, 7, 8, 9] --> another chunk
print(next(x)) #[10, 11, 12, 13, 14]..and so on
```
