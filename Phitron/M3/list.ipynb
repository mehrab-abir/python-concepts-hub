### List Comprehnsion

```python
random_list_1 = [x for x in range(1,10)]
print(random_list_1)

random_list_2 = [x*2 for x in range(1,20)]
print(random_list_2)

# even numbrrs from 0 to 20
even_numbers = [i for i in range(0,21) if i%2==0]
print(even_numbers)
```
