### Lowest Number and Position

````python
size = int(input("How many numbers: "))
nums = list(map(int,input().split()))

lowest = nums[0]
pos = 1

for i in range(size):
    if(nums[i] < lowest):
        lowest = nums[i]
        pos = i + 1
        
print(nums)
        
print(lowest, pos)
````

### Swap minimum number with maximum number (Python).

````python
n = int(input("Enter size: "))
nums = list(map(int, input("Enter Numbers: ").split()))

print(nums)

minimum = nums[0]

for num in nums:
    if(num < minimum):
        minimum = num

maximum = nums[0]

for num in nums:
    if(num > maximum):
        maximum = num

for num in nums:
    if(num == minimum):
        num = maximum
    elif(num == maximum):
        num = minimum
    print(num,end=" ")
````

### Check whether a string contains 101 or 010

````python
t = int(input("Number of test cases: "))

for i in range(t):
    s = input("Enter string: ")

    found = False

    if("010" in s or "101" in s):
        found = True
        
    if(found):
        print("Good")
    else:
        print("Bad")
````

		string subStr = str.substr(i, 5);

		}
