Over-write file

```python
# overwrite file - erase exisitng content and writing new content only
content = input("Write your content: ");

with open('./newFile.txt','w') as file:
    file.write(content)
    file.write('\n')

print("File over-written")
```

Append content to a file

```python
newContent = input("Write your new content: ");

with open('./newFile.txt','a') as file:
    file.write(newContent)
    file.write('\n')

print("New content added")

# reading file
print("\nFile content: \n")
with open('./newFile.txt','r') as file:
    print(file.read())
```
