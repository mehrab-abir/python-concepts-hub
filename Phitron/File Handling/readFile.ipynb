when file is in the same directory

```python
file = open('sample.txt','r') # 'r' means --> open in 'read' mode

content = file.read() # then, actually reading the file

file.close()

print(content)
```

when file is in a child directory

```python
file = open('./ChildFiles/maple.txt','r')

content = file.read()

file.close()

print(content)
```

when file is in an external directory

```python
file = open('../Random Files/random.txt')
content = file.read()

file.close()

print(content)

# print(file.closed) #True/False
```

'with' keyword -- file automatically closes after read

```python
with open('./sample.txt','r') as file:
    content = file.read()
print(content)
    
# print(file.closed) # True
```

reading line by line

```python
with open('./sample.txt','r') as file:
    for line in file:
        # print(line) # --> prints the newline casued by 'print()' as well
        print(line.strip()) # --> removes the extra new line casued by 'print()'
```

```python
with open('./sample.txt','r') as file:
    print(file.readlines())
    
    # see the next cell below
```

```python
with open('./sample.txt','r') as file:
    print(list(map(str.strip,file.readlines())))
    # str.strip --> to remove the trailing '\n'
    # then converting the object returned by 'map()' to a list
```
