when file is in the same directory

```python
file = open('sample.txt','r')

content = file.read()

file.close()

print(content)
```

when file is in a child directory

```python
file = open('./ChildFiles/maple.txt','r')

content = file.read()

file.close()

print(content)
```

when file is in an external directory

```python
file = open('../Random Files/random.txt')
content = file.read()

file.close()

print(content)

# print(file.closed) #True/False
```

'with' keyword -- file automatically closes after read

```python
with open('./sample.txt') as file:
    content = file.read()
    print(content)
    
# print(file.closed) # True
```
