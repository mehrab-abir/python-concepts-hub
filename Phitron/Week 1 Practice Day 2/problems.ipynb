Create and Write to a File


```python
def create_file(filename, content):
    with open(filename,'w') as file:
        file.write(content)
        
content = "Hello, this is a new file and its content"   
create_file("fileA.txt", content)
```

Read lines with line numbers

```python
def read_lines_with_numbers(filename):
    line_number = 1
    
    with open(filename) as file:
        for line in file:
            print(f"{line_number}: {line.strip()}")
            line_number += 1

read_lines_with_numbers("fileA.txt")
```

Append log entries to a file

```python
def add_log(log_message):
    with open('./log.txt','a') as file:
        file.write(f"{log_message}\n")

# add_log("hello")
# add_log("user created")
add_log("file uploaded")

print("End!")
```

  Read from a Specific File Pointer Position

```python
def read_from_position(filename, position):
    with open(filename,'r') as file:
        file.seek(position)
        print(file.read())
        
read_from_position('log.txt',5)
```

file read with exception handling

```python
def safe_read(filename):
    try:
        with open(filename,'r') as file:
            print(file.read())
    except Exception as e:
        print(e)
    finally:
        print("Operation completed")
        
safe_read("log.txt")
```
