```python
# count the number of even, odd, positive and negative number

x = int(input("How many numbers: "))
numbers = list(map(int,input("Enter numbers: ").split()))

print(numbers)

even = 0
odd = 0
positive = 0
negative = 0

for num in numbers:
    if(num > 0):
        positive += 1
    if(num < 0):
        negative += 1
    if(num % 2 == 0):
        even += 1
    if(num %2 == 1):
        odd += 1
        
print("Even: ",even)
print("Odd: ",odd)
print("Positive: ",positive)
print("Negative: ",negative)
```

```python
## print digits from right to left

num = int(input("Enter a number: "))
print("The number: ", num)

if(num == 0):
    print(num)
else:
    while num != 0:
            digit = num % 10
            print(digit,end=" ")
            num = int(num / 10)
            # num //=10 # same as the line above
```
