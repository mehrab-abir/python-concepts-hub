```python
## maximum number from a list

nums = list(map(int,input("Enter numbers: ").split()))

maximum = nums[0]

for i in range(len(nums)-1):
    if(nums[i] > maximum):
        maximum = nums[i]
        
print(f"Maximum number: {maximum}")
```

```python
## maximum number from a list

## more Pythonic way

nums = list(map(int,input("Enter numbers: ").split()))

maximum = nums[0]

for num in nums:
    if(num > maximum):
        maximum = num
        
print(f"Maximum number is: {maximum}")
```
