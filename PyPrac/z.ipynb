```python
nums =input("Enter numbers: ") #type will be string
print(nums)

print(type(nums))
```

```python
nums =input("Enter numbers: ").split() #type will be list(array),,but each item is a string,,e.g. ['56','12']
print(nums)

print(type(nums))
```

```python
numbers =list(map(int,input("Enter numbers: ").split())) #type will be list(array),,and each item is an int number..e.g. [56,12]
print("The numbers:",numbers)

print(type(numbers))
```
